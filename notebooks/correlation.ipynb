```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('../data/processed/final_data.csv')
```

```python
data.head()
```

```python
# Select 70% of unique users
# Select 70% of unique users
sampled_users = data['user_id'].drop_duplicates().sample(
    frac=0.70,
    random_state=42
)

# Keep all rows belonging to selected users
sampled_df = data[data['user_id'].isin(sampled_users)].copy()

# Check dimensions
print("Original shape:", data.shape)
print("Sampled shape:", sampled_df.shape)
print("Unique users:", sampled_df['user_id'].nunique())
```

```python
c1 = pd.crosstab(data['education'],data['buy'],margins=True)
c1
```

```python
from scipy.stats import chi2_contingency
chi2_stat,p_value,dof,expected = chi2_contingency(c1)
print(chi2_stat)
print(p_value)
```
